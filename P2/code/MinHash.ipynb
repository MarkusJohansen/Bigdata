```python
Input_Matrix = [] # Create an empty array of input matrix
```

```python
## Compute the values for the input matrix here ##
for sentence in [Sentence_1, Sentence_2, Sentence_3, Sentence_4]:
    row = [0] * len(unique_words)
    words = sentence.lower().split()
    for word in words:
        if word in unique_words:
            row[unique_words.index(word)] = 1
    Input_Matrix.append(row)

Input_Matrix
```

```python
Sig_M = [] # Create a MinHash signature matrix array with inf values
for i in range(3):
    Sig_M.append([np.inf] * 4)

Sig_M
```

```python
# Generate hash functions
```

```python
hash_1 = lambda x, a=1, b=3: (a * x + b) % 10 + 1
```

```python
hash_2 = lambda x, a=7, b=1: (a * x + b) % 10 + 1
```

```python
hash_3 = lambda x, a=7, b=2: (a * x + b) % 10 + 1
```

```python
hash_list = [hash_1, hash_2, hash_3]
```

```python
hash_list
```

```python
## Using the hash functions and the input matrix compute the MinHash signature matrix here ##
```

```python
# print the updated MinHash signature matrix

# For each column (feature)
for col_id in range(len(Input_Matrix[0])):
    # For each row (document)
    for row_id, row in enumerate(Input_Matrix):
        # If this feature is present in the document
        if row[col_id] == 1:
            # Update signature with minimum hash value
            for h_id, hash_func in enumerate(hash_list):
                hash_value = hash_func(col_id + 1)

                if hash_value < Sig_M[h_id][row_id]:
                    Sig_M[h_id][row_id] = hash_value

Sig_M
```
